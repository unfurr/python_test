```python
#格式化输出，%s是字符串，%d是十进制整数%c是字符%f是浮点数
print('我的名字是%s,年龄是%d' % ("01",18))#注意%（）前面不能有逗号
print('我的名字是{}，年龄是{}'.format('01',18))#注意format前面有点
#第三种,只用改变一种
name,age = "01",18
print(f"我的名字是{name},年龄是{age}")#注意前面这个f是不能丢掉的
```

```python
#string字符类型，字符串索引正着数是从0开始，倒着数是从-1开始往下降
str1 = "hello world"#单双三引号都可以使用，三引号“”“可以当多行注释
print(type(str1))
print("""
a
b
c
""")#多行输出
tri2 = "a"
tri3 = "b"
print(tri2+tri3)#字符串的拼接
```

```python
#字符串的查询
str1 = "hello world"
print(str1[2])
print(str1[1:6:2])#切片格式[strat:end:step],不写就是省略，注意是左闭右开,如果step是负数那就是从后向前
print(len(str1))
```

```python
#list列表
my = [1,"2",[1,2,3]]#索引位置和之前一阿姨那个，第一个是0，最后一个是-1
print(my[-1])
print(my[1:3:1])#切片和之前一样
my1 = [1,2,3]
print(my+my1)
```

```python
a = list(range(0,10,1))#range和其他切片一样也会可以写成range(0,10),默认值不写
print(a)
a.append(10)#增
a.remove(5)#删
a[0]=11#改
a[0:3]#查，和切片一样
```

```python
a = list(range(0,10,1))#range和其他切片一样也会可以写成range(0,10),默认值不写
print(a)
a.append(10)#增
a.remove(5)#删
a[0]=11#改
a[0:3]#查，和切片一样
print(max(a))
print(min(a))
print(sum(a))
a.sort()
print(a)#升序
a.sort(reverse=True)
print(a)#降序

```

```python
#元组，不可以修改，并且元组中的元素类型相同
b = (1,2,1,4)#索引都相同,切片也一样
print(b[0])
print(b.count(1))#计数，计算1出现次数
b.index(2)#2最先出现的位置
```
