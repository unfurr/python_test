```python
#类和对象
#类是具有相同属性和共性的一类事如动物
#对象是类的具体体现如猫狗猪
#自定义类class 类名（）：
    #方法属性
class Pig():#首字母尽量大写
    pass#用pass占位，避免语法错误
class Person():
    #类属性
    name = "li"
    leg = 2
    arm = 2
    #以下函数叫方法
    def smart(self):#self不能丢,这只是一个形参，可以写成其他形式但不能没有,有两个形参的时候第一个永不用写，因为self就是对象，所以调用时都是把实参传到第二个形参
        print("True")
    def walk(self):
        print("True")
people = Person()#people就是Person的一个对象，此时people这个对象就有了Person这个类中所有定义,此过程中叫类的实例化,1、不同实例化内存空间不同
zhu = Pig()
people.smart()
print(people.arm)#相当于封装，封装了相关函数和数据
```

```python
#构造方法 __init__
class Person():
    def init(self,name,age,height):
        self.name = name
        self.age = age
        self.height = height
    def smart(self):
        print(f"{self.name} is smart")
    def walk(self):
        print("True")
people3 = Person()
people3.init("li","12","160")
people3.smart()
#以上都是引子
```

```python
#__init__函数的使用
class Person():
    def __init__(self,name,age,height):
        print("自动执行")
        self.name = name
        self.age = age
        self.height = height
    def smart(self):
        print(f"{self.name} is smart")
    def walk(self):
        print("True")
people3 = Person("li","12","160")#使用了直接初始化
people3.smart()
```

```python
#类方法
class Person():
    num = 0
    def __init__(self,name,age,height):
        print("自动执行")
        self.name = name
        self.age = age
        self.height = height
        Person.num +=1#注意此时是类对象，如果写成num，就变成了局部变量
    def smart(self):
        print(f"{self.name} is smart")
    def walk(self):
        print("True")
    @classmethod
    def show_num(cls):
        print(f"当前人数是{Person.num}")
Person.show_num()
people5 = Person("li","12","160")#使用了直接初始化
people5.smart()
```
