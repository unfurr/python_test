```python
f=open("try.txt","a")#w是写入会覆盖院文件，a是追加模式
print("hello world",file=f)
f.close()#关闭就是保存
```

```python
with open("try.txt","a") as f:#使用with语句可以打开文件,而且是自动关闭，如果没有这个该文件的话会创建这个文件
    print("hello world",file=f)
```

```python
#格式化输出，%s是字符串，%d是十进制整数%c是字符%f是浮点数
print('我的名字是%s,年龄是%d' % ("01",18))#注意%（）前面不能有逗号
print('我的名字是{}，年龄是{}'.format('01',18))#注意format前面有点
#第三种,只用改变一种
name,age = "01",18
print(f"我的名字是{name},年龄是{age}")#注意前面这个f是不能丢掉的
```

```python
#输入input函数
user_name = input("请你输入姓名：")
print('你的名字是:', user_name)#注意逗号后面又空格
print(type(user_name))
age = int(input('请输入你的年龄：'))
print("你的年龄是：", age)
print(type(age))#变量类型
print(id(age))#age的内存地址
```
